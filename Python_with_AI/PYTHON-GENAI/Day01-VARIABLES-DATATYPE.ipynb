```python
# ================================================
# ✅ GOOD variable names vs ❌ BAD variable names
# ================================================

# ✅ GOOD - clear, readable, uses underscore for spaces
player_name      = "MS Dhoni"
total_runs       = 10773
matches_played   = 350
is_wicket_keeper = True

# ❌ BAD - these will cause errors or confusion
# 1player = "Dhoni"     ❌ Cannot START with a number
# player name = "Dhoni" ❌ No SPACES allowed
# player-name = "Dhoni" ❌ No HYPHENS allowed

# ⚠️ Technically works BUT confusing - avoid
x    = "MS Dhoni"   # What is x? Makes no sense
data = 10773        # data of what?
n    = 350          # n means nothing

# ✅ Python naming convention: use_underscore_for_spaces
# This is called "snake_case" – the Python standard
```
